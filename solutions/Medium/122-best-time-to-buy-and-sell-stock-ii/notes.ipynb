to calculate general profit with multiple transactions i assume to buy cheap and sell on highs, when the price is going up, to maximize profit with multiple transactions

```python
from typing import List

class Solution:
    def maxProfit(self, prices: List[int]) -> int:
        # if the prices list is empty, then we just return 0 since there is no profit to be made
        if not prices:
          return 0

        # variable to keep track of total profit
        profit = 0

        # for the prices starting from the first index (second day), I just check if
        # there is a profit to be made by comparing the current price with the previous price. 
        # If there is a profit, I add it to the total profit.
        for i in range(1, len(prices)):
          # if the current price is greater than the previous price, then there is a profit
          if prices[i] > prices[i - 1]:
            profit += prices[i] - prices[i-1]
        
        return profit


```

the first idea was to calculate overall biggest profit during the presented days. But it was focused on doing just one trade, that has max profit

```python
from typing import List

class Solution:
    def maxProfit(self, prices: List[int]) -> int:
        if not prices:
          return 0

        maxProfit: int = 0

        for i in range(len(prices) - 1):
          j = i + 1

          while j < len(prices):
            # our profit 
            profit = prices[j] - prices[i]

            if profit > maxProfit:
              maxProfit = profit
            
            j += 1

        return maxProfit
```
